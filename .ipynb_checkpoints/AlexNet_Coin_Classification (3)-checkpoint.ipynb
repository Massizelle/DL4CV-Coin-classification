## 0. Imports et configuration

````python
import os
import copy
import time
import warnings
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image, ImageFile

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
from torchvision import models

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, f1_score, top_k_accuracy_score

from tqdm.notebook import tqdm

warnings.filterwarnings('ignore')
ImageFile.LOAD_TRUNCATED_IMAGES = True

DATA_DIR  = Path("./coins") 
TRAIN_DIR = DATA_DIR / "train"
TEST_DIR  = DATA_DIR / "test"

# Vérification que les données sont bien là
for p in [DATA_DIR / "train.csv", TRAIN_DIR, TEST_DIR]:
    assert p.exists(), f"Chemin introuvable : {p}"

# Hyperparamètres
IMG_SIZE   = 224
BATCH_SIZE = 64
NUM_EPOCHS = 30
LR         = 1e-3
SEED       = 42

torch.manual_seed(SEED)
np.random.seed(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device : {DEVICE}")
if torch.cuda.is_available():
    print(f"GPU    : {torch.cuda.get_device_name(0)}")
    print(f"VRAM   : {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
````

## 1. Exploration du dataset (EDA)

````python
train_df = pd.read_csv(DATA_DIR / "train.csv")
test_df  = pd.read_csv(DATA_DIR / "test.csv")
sub_df   = pd.read_csv(DATA_DIR / "sample_submission.csv")

print(f"Train : {len(train_df)} images, {train_df['class'].nunique()} classes")
print(f"Test  : {len(test_df)} images")
print()
train_df.head()
````

````python
# Décomposition de la colonne 'class' en denomination / currency / country
def parse_class(c):
    parts = [s.strip() for s in str(c).split(',')]
    denomination = parts[0] if len(parts) > 0 else ''
    currency     = parts[1] if len(parts) > 1 else ''
    country      = parts[2] if len(parts) > 2 else ''
    return denomination, currency, country

parsed = train_df['class'].apply(parse_class)
train_df['denomination'] = [x[0] for x in parsed]
train_df['currency']     = [x[1] for x in parsed]
train_df['country']      = [x[2] for x in parsed]

class_counts = train_df['class'].value_counts()

print("Distribution des images par classe :")
print(f"  Min  : {class_counts.min()}")
print(f"  Max  : {class_counts.max()}")
print(f"  Moy  : {class_counts.mean():.1f}")
print(f"  Méd  : {class_counts.median():.1f}")
print(f"  Classes avec < 10 images : {(class_counts < 10).sum()}")
````

## 2. Prétraitement et DataLoaders

````python
# Encodage des labels
le = LabelEncoder()
train_df['label'] = le.fit_transform(train_df['class'])
NUM_CLASSES = len(le.classes_)
print(f"{NUM_CLASSES} classes")

# Split stratifié train / val (85% / 15%)
train_split, val_split = train_test_split(
    train_df, test_size=0.15, stratify=train_df['label'], random_state=SEED
)
print(f"Train : {len(train_split):,}  |  Val : {len(val_split):,}")
````

````python
# Normalisation ImageNet (AlexNet a été pré-entraîné dessus)
MEAN = [0.485, 0.456, 0.406]
STD  = [0.229, 0.224, 0.225]

# Augmentation pour l'entraînement
# On utilise RandomCrop après un Resize(256) : c'est la stratégie d'origine d'AlexNet
train_transforms = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomCrop(224),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=15),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.05),
    transforms.RandomGrayscale(p=0.05),
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD),
])

val_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD),
])

print("Transforms définis.")
````

````python
class CoinDataset(Dataset):
    def __init__(self, df, img_dir, transform=None, has_labels=True):
        self.df         = df.reset_index(drop=True)
        self.img_dir    = Path(img_dir)
        self.transform  = transform
        self.has_labels = has_labels
        self.exts       = ['.jpg', '.jpeg', '.png', '.webp']

    def __len__(self):
        return len(self.df)

    def _find_image(self, img_id):
        for ext in self.exts:
            p = self.img_dir / (str(img_id) + ext)
            if p.exists():
                return p
        return None

    def __getitem__(self, idx):
        row      = self.df.iloc[idx]
        img_path = self._find_image(row['id'])

        if img_path is None:
            # image manquante -> gris neutre (cas rare)
            img = Image.new('RGB', (224, 224), color=(128, 128, 128))
        else:
            try:
                img = Image.open(img_path).convert('RGB')
            except Exception:
                img = Image.new('RGB', (224, 224), color=(128, 128, 128))

        if self.transform:
            img = self.transform(img)

        if self.has_labels:
            return img, int(row['label'])
        return img, str(row['id'])


# num_workers=0 pour éviter les problèmes sur Windows/Mac local
# (mettre 2 ou 4 sur Linux si besoin de perf)
NW = 0 if os.name == 'nt' else 2

train_dataset = CoinDataset(train_split, TRAIN_DIR, transform=train_transforms)
val_dataset   = CoinDataset(val_split,   TRAIN_DIR, transform=val_transforms)
test_dataset  = CoinDataset(test_df,     TEST_DIR,  transform=val_transforms, has_labels=False)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True,
                          num_workers=NW, pin_memory=torch.cuda.is_available())
val_loader   = DataLoader(val_dataset,   batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=NW, pin_memory=torch.cuda.is_available())
test_loader  = DataLoader(test_dataset,  batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=NW, pin_memory=torch.cuda.is_available())

print(f"Train batches : {len(train_loader)}")
print(f"Val   batches : {len(val_loader)}")
print(f"Test  batches : {len(test_loader)}")
````

## 3. Architecture AlexNet

AlexNet (Krizhevsky et al., 2012) est composé de 5 couches convolutives suivies de 3 couches fully-connected. Par rapport à l'architecture originale :
- On remplace la LRN (Local Response Normalization) par de la BatchNorm (optionnelle) car c'est plus stable
- On utilise un AdaptiveAvgPool pour s'abstraire de la taille d'entrée exacte
- L'initialisation des poids suit Kaiming (He et al., 2015) qui est mieux adaptée aux ReLU

```
Input (3×224×224)
  → Conv1(96, 11×11, s=4) → ReLU → MaxPool(3×3, s=2)
  → Conv2(256, 5×5, p=2)  → ReLU → MaxPool(3×3, s=2)
  → Conv3(384, 3×3, p=1)  → ReLU
  → Conv4(384, 3×3, p=1)  → ReLU
  → Conv5(256, 3×3, p=1)  → ReLU → MaxPool(3×3, s=2)
  → AdaptiveAvgPool(6×6) → Flatten(9216)
  → FC(4096) → ReLU → Dropout(0.5)
  → FC(4096) → ReLU → Dropout(0.5)
  → FC(NUM_CLASSES)
```

````python
class AlexNetCustom(nn.Module):
    """
    AlexNet re-implémenté pour la classification de pièces.
    use_bn=True ajoute BatchNorm après chaque conv (améliore la convergence).
    """
    def __init__(self, num_classes=315, use_bn=False, dropout=0.5):
        super().__init__()

        def conv_block(in_c, out_c, k, s=1, p=0):
            layers = [nn.Conv2d(in_c, out_c, kernel_size=k, stride=s, padding=p)]
            if use_bn:
                layers.append(nn.BatchNorm2d(out_c))
            layers.append(nn.ReLU(inplace=True))
            return layers

        self.features = nn.Sequential(
            *conv_block(3,   96,  11, s=4, p=2),
            nn.MaxPool2d(kernel_size=3, stride=2),
            *conv_block(96,  256,  5, p=2),
            nn.MaxPool2d(kernel_size=3, stride=2),
            *conv_block(256, 384,  3, p=1),
            *conv_block(384, 384,  3, p=1),
            *conv_block(384, 256,  3, p=1),
            nn.MaxPool2d(kernel_size=3, stride=2),
        )

        self.avgpool = nn.AdaptiveAvgPool2d((6, 6))

        self.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(256 * 6 * 6, 4096),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout),
            nn.Linear(4096, 4096),
            nn.ReLU(inplace=True),
            nn.Linear(4096, num_classes),
        )

        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                nn.init.normal_(m.weight, 0, 0.01)
                nn.init.constant_(m.bias, 0)

    def forward(self, x):
        x = self.features(x)
        x = self.avgpool(x)
        x = torch.flatten(x, 1)
        return self.classifier(x)


# Vérification rapide des dimensions
m = AlexNetCustom(num_classes=NUM_CLASSES, use_bn=False)
dummy = torch.zeros(2, 3, 224, 224)
out   = m(dummy)
n_params = sum(p.numel() for p in m.parameters() if p.requires_grad)
print(f"AlexNet from scratch — {n_params:,} paramètres")
print(f"Output shape : {out.shape}")
````

````python
def make_alexnet_pretrained(num_classes, freeze_features=False):
    """
    AlexNet pré-entraîné ImageNet, tête adaptée pour num_classes.
    freeze_features=True : on gèle les conv et on entraîne uniquement le classifier.
    freeze_features=False : fine-tuning complet du réseau.
    """
    model = models.alexnet(weights=models.AlexNet_Weights.IMAGENET1K_V1)

    if freeze_features:
        for param in model.features.parameters():
            param.requires_grad = False

    # Remplace la dernière couche : 1000 classes ImageNet -> NUM_CLASSES
    in_features = model.classifier[6].in_features
    model.classifier[6] = nn.Linear(in_features, num_classes)

    return model


model_pretrained = make_alexnet_pretrained(NUM_CLASSES, freeze_features=False)
total_params     = sum(p.numel() for p in model_pretrained.parameters())
trainable_params = sum(p.numel() for p in model_pretrained.parameters() if p.requires_grad)
print(f"AlexNet pretrained — {total_params:,} paramètres totaux ({trainable_params:,} entraînables)")
````

## 4. Entraînement

````python
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()
    total_loss = 0.0
    correct    = 0
    total      = 0

    for imgs, labels in tqdm(loader, desc="  train", leave=False):
        imgs, labels = imgs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(imgs)
        loss    = criterion(outputs, labels)
        loss.backward()
        # Gradient clipping : évite les explosions de gradient
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        total_loss += loss.item() * imgs.size(0)
        preds       = outputs.argmax(1)
        correct    += (preds == labels).sum().item()
        total      += imgs.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion, device, top_k=(1, 5)):
    model.eval()
    total_loss = 0.0
    total      = 0
    correct    = {k: 0 for k in top_k}

    for imgs, labels in tqdm(loader, desc="  val  ", leave=False):
        imgs, labels = imgs.to(device), labels.to(device)
        outputs = model(imgs)
        loss    = criterion(outputs, labels)
        total_loss += loss.item() * imgs.size(0)
        total      += imgs.size(0)

        for k in top_k:
            _, pred = outputs.topk(k, dim=1)
            correct[k] += pred.eq(labels.view(-1, 1).expand_as(pred)).sum().item()

    return {"loss": total_loss / total, **{f"top{k}": correct[k] / total for k in top_k}}


def train_model(model, train_loader, val_loader, num_epochs, lr, device,
                model_name="alexnet", patience=5):
    model = model.to(device)

    # Label smoothing : réduit la sur-confiance du modèle
    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
    optimizer = optim.SGD(model.parameters(), lr=lr, momentum=0.9, weight_decay=5e-4)
    # Cosine annealing : LR décroît en cosinus jusqu'à eta_min
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=num_epochs, eta_min=1e-5)

    history  = {'train_loss': [], 'train_acc': [], 'val_loss': [],
                'val_top1': [], 'val_top5': [], 'lr': []}
    best_acc = 0.0
    best_wts = copy.deepcopy(model.state_dict())
    no_improve = 0

    print(f"\nEntraînement : {model_name}")
    print(f"Epochs={num_epochs} | LR={lr} | Batch={BATCH_SIZE} | Device={device}\n")

    for epoch in range(1, num_epochs + 1):
        t0 = time.time()

        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, device)
        val_metrics           = evaluate(model, val_loader, criterion, device)
        scheduler.step()

        current_lr = optimizer.param_groups[0]['lr']
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_metrics['loss'])
        history['val_top1'].append(val_metrics['top1'])
        history['val_top5'].append(val_metrics['top5'])
        history['lr'].append(current_lr)

        elapsed = time.time() - t0
        is_best = val_metrics['top1'] > best_acc

        if is_best:
            best_acc   = val_metrics['top1']
            best_wts   = copy.deepcopy(model.state_dict())
            torch.save(best_wts, f'{model_name}_best.pth')
            no_improve = 0
            flag = "[best]"
        else:
            no_improve += 1
            flag = f"(+{no_improve}/{patience})"

        print(f"[{epoch:02d}/{num_epochs}] "
              f"loss={train_loss:.4f}/{val_metrics['loss']:.4f}  "
              f"acc={train_acc:.3f}/{val_metrics['top1']:.3f}  "
              f"top5={val_metrics['top5']:.3f}  "
              f"lr={current_lr:.2e}  {elapsed:.0f}s  {flag}")

        if no_improve >= patience:
            print(f"Early stopping à l'époque {epoch}")
            break

    print(f"\nMeilleure val Top-1 : {best_acc*100:.2f}%")
    model.load_state_dict(best_wts)
    return model, history, best_acc
````

````python
# Choix du modèle
# USE_PRETRAINED = True  -> AlexNet ImageNet fine-tuning (recommandé)
# USE_PRETRAINED = False -> AlexNet from scratch

USE_PRETRAINED = True

if USE_PRETRAINED:
    model      = make_alexnet_pretrained(NUM_CLASSES, freeze_features=False)
    model_name = "alexnet_pretrained"
    lr_choice  = 1e-3
    print("AlexNet pretrained (fine-tuning complet)")
else:
    model      = AlexNetCustom(num_classes=NUM_CLASSES, use_bn=True, dropout=0.5)
    model_name = "alexnet_scratch"
    lr_choice  = 1e-2
    print("AlexNet from scratch")

model, history, best_top1 = train_model(
    model        = model,
    train_loader = train_loader,
    val_loader   = val_loader,
    num_epochs   = NUM_EPOCHS,
    lr           = lr_choice,
    device       = DEVICE,
    model_name   = model_name,
)
````

````python
def plot_training_curves(history, title="AlexNet"):
    epochs = range(1, len(history['train_loss']) + 1)
    fig, axes = plt.subplots(1, 3, figsize=(16, 4))

    axes[0].plot(epochs, history['train_loss'], label='Train')
    axes[0].plot(epochs, history['val_loss'],   label='Val')
    axes[0].set_title('Loss')
    axes[0].set_xlabel('Époque')
    axes[0].legend()
    axes[0].grid(alpha=0.3)

    axes[1].plot(epochs, [x*100 for x in history['train_acc']],  label='Train Acc')
    axes[1].plot(epochs, [x*100 for x in history['val_top1']],   label='Val Top-1')
    axes[1].plot(epochs, [x*100 for x in history['val_top5']],   label='Val Top-5', linestyle='--')
    axes[1].set_title('Accuracy (%)')
    axes[1].set_xlabel('Époque')
    axes[1].legend()
    axes[1].grid(alpha=0.3)

    axes[2].semilogy(epochs, history['lr'])
    axes[2].set_title('Learning Rate (Cosine Annealing)')
    axes[2].set_xlabel('Époque')
    axes[2].grid(alpha=0.3)

    best_v = max(history['val_top1'])
    best_e = history['val_top1'].index(best_v) + 1
    plt.suptitle(f'{title} — Best Val Top-1 : {best_v*100:.2f}% (époque {best_e})',
                 fontsize=12, fontweight='bold')
    plt.tight_layout()
    plt.savefig('training_curves.png', dpi=150, bbox_inches='tight')
    plt.show()

plot_training_curves(history, title="AlexNet Pretrained" if USE_PRETRAINED else "AlexNet Scratch")
````

## 5. Évaluation quantitative et comparaison au leaderboard

````python
@torch.no_grad()
def get_predictions(model, loader, device):
    model.eval()
    all_preds, all_labels, all_probs = [], [], []

    for imgs, labels in tqdm(loader, desc="Inférence"):
        imgs   = imgs.to(device)
        logits = model(imgs)
        probs  = torch.softmax(logits, dim=1).cpu().numpy()
        preds  = logits.argmax(1).cpu().numpy()
        all_probs.extend(probs)
        all_preds.extend(preds)
        all_labels.extend(labels.numpy())

    return np.array(all_preds), np.array(all_labels), np.array(all_probs)


preds, labels, probs = get_predictions(model, val_loader, DEVICE)

top1 = (preds == labels).mean()
top5 = top_k_accuracy_score(labels, probs, k=5)

print("=" * 55)
print("  Résultats sur le validation set")
print("=" * 55)
print(f"  Top-1 Accuracy : {top1*100:.2f}%")
print(f"  Top-5 Accuracy : {top5*100:.2f}%")
print()
print("  Comparaison Leaderboard Kaggle")
print(f"  AlexNet (notre) : {top1*100:.2f}%")
print(f"  Leaderboard #1  : ~90.17%")
print(f"  Leaderboard #2  : ~90.17%")
print(f"  Leaderboard #3  : ~89.55%")
print("=" * 55)
print()
print("Note : notre évaluation porte sur le split de validation (15% du train),")
print("alors que le leaderboard Kaggle est calculé sur 50% du test set officiel.")
````

````python
# Rapport de classification sur les 20 premières classes
unique_labels = np.unique(labels)
class_names   = le.inverse_transform(unique_labels)[:20]
print(classification_report(
    labels, preds,
    labels=unique_labels[:20],
    target_names=[c[:30] for c in class_names],
    zero_division=0
))
````

````python
# Matrice de confusion normalisée sur les 15 classes les plus fréquentes
top15_labels = [x[0] for x in Counter(labels).most_common(15)]
mask         = np.isin(labels, top15_labels)
cm           = confusion_matrix(labels[mask], preds[mask], labels=top15_labels)
cm_norm      = cm.astype(float) / cm.sum(axis=1, keepdims=True)
class_names_15 = [le.inverse_transform([l])[0].split(',')[0][:12] for l in top15_labels]

plt.figure(figsize=(12, 10))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Blues',
            xticklabels=class_names_15, yticklabels=class_names_15,
            linewidths=0.5, linecolor='#e0e0e0')
plt.title('Matrice de confusion normalisée (Top 15 classes)', fontweight='bold')
plt.ylabel('Classe réelle')
plt.xlabel('Classe prédite')
plt.xticks(rotation=45, ha='right', fontsize=9)
plt.yticks(rotation=0, fontsize=9)
plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
````

````python
# F1-score par classe : meilleures et pires performances
f1_per_class = f1_score(labels, preds, average=None, zero_division=0,
                        labels=np.arange(NUM_CLASSES))
support      = np.bincount(labels, minlength=NUM_CLASSES)

results_df = pd.DataFrame({
    'class':   le.inverse_transform(np.arange(NUM_CLASSES)),
    'f1':      f1_per_class,
    'support': support,
}).query('support > 0').sort_values('f1', ascending=False)

print("Top 10 classes (meilleur F1) :")
print(results_df.head(10)[['class', 'f1', 'support']].to_string(index=False))
print()
print("Top 10 classes (pire F1) :")
print(results_df.tail(10)[['class', 'f1', 'support']].to_string(index=False))
````

## 7. Génération de la soumission Kaggle

````python
@torch.no_grad()
def predict_test(model, loader, device):
    model.eval()
    all_ids, all_preds = [], []

    for imgs, ids in tqdm(loader, desc="Test inference"):
        imgs   = imgs.to(device)
        logits = model(imgs)
        preds  = logits.argmax(1).cpu().numpy()
        all_ids.extend(ids)
        all_preds.extend(preds)

    return all_ids, all_preds


test_ids, test_preds = predict_test(model, test_loader, DEVICE)

pred_classes  = le.inverse_transform(test_preds)
submission_df = pd.DataFrame({'id': test_ids, 'class': pred_classes})
submission_df.to_csv('submission_alexnet.csv', index=False)

print(f"Soumission générée : submission_alexnet.csv ({len(submission_df)} prédictions)")
submission_df.head(10)
````

## 8. Critique et axes d'amélioration

### Limites d'AlexNet pour ce problème

AlexNet date de 2012 et présente plusieurs faiblesses sur un problème à 315 classes fine-grained comme les pièces de monnaie :

- **Profondeur limitée** : 5 couches convolutives seulement. Les architectures modernes (ResNet, EfficientNet) vont bien au-delà et capturent des représentations plus riches.
- **Absence de skip connections** : sans connexions résiduelles, les gradients se dégradent sur les premières couches (vanishing gradient). Les ResNets règlent ce problème.
- **Sur-paramétrage des FC** : les couches FC (4096→4096) représentent ~80% des paramètres mais sont moins efficaces que des couches de convolution profondes.
- **Déséquilibre de classes** : le dataset présente un fort déséquilibre. AlexNet avec CrossEntropy standard a tendance à ignorer les classes rares.

### Pistes d'amélioration

| Axe | Idée | Gain estimé |
|-----|------|-------------|
| **Architecture** | EfficientNetV2-S ou ConvNeXt-T en fine-tuning | +10–15% acc |
| **Augmentation** | Mixup, CutMix, RandAugment | +3–5% acc |
| **Classes déséquilibrées** | Focal Loss ou weighted sampler | améliore les classes rares |
| **Scheduler** | Warmup cosine (progressive LR) | convergence plus stable |
| **Inférence** | Test-Time Augmentation (TTA) | +1–2% acc |
| **Ensemble** | Combiner AlexNet + ResNet50 (voting) | +2–3% acc |

En particulier, EfficientNetV2 fine-tuné depuis ImageNet-21k devrait permettre d'atteindre les scores du top du leaderboard (~90%).
